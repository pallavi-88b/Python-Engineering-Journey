## Question 06: Take a string as input and count the number of uppercase and lowercase in it.

```python
word = input("Enter String: ")
lowercase = 0
uppercase = 0
for char in word:
    if char.islower():
        lowercase = lowercase + 1
    elif char.isupper():
        uppercase = uppercase + 1
print(f' The number of lowercase :{lowercase}')
print(f' The number of uppercase :{uppercase}')
```

## Question 16: Take a string as input and find the character that occurs most frequently.

```python
word = input("Enter string:")
freq = ""
max_count = 0
for char in word.lower():
    if char == " ":
        continue
    current_count = word.count(char) 
    if current_count > max_count:
       max_count = current_count
       freq = char
print(freq)         
         
     
```

## Question 19: Take a string as input and check whether they are anagrams of each other.

```python
word_1 = input("Enter string:")
word_2 = input("Enter string:")
freq = ""
freq_2 = ""
for char in word_1:
    if char in freq:
        continue 
    else:
        freq = freq + char
        
        
for char in word_2:
    if char in freq_2:
        continue
    else:
        freq_2 = freq_2 + char
        
        
if  len(word_1) != len(word_2):
        print("Not anagrams")
else:
        
        for char in freq :
            if word_1.count(char) != word_2.count(char):
                print("Not anagrams")
                break
        else:
            print("anagrams")
        
```
