```python
numbers = [10,34,56,79,38]
def outer(numbers):
    def inner():
        low = numbers[0]
        for num in numbers:
            if num < low:
                low = num
        return low
    return inner()
outer(numbers)
```

```python
numbers = [-10,34,-56,79,38]
def outer(numbers):
    def inner():
        high = numbers[0]
        for num in numbers:
            if num > high and num > 0:
                high = num
        return high
    return inner()
outer(numbers)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟐𝟎 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙤𝙪𝙩𝙚𝙧(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣. 𝙄𝙣𝙨𝙞𝙙𝙚 𝙞𝙩, 𝙘𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙞𝙣𝙣𝙚𝙧() 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨: 𝙏𝙤𝙩𝙖𝙡 𝙣𝙪𝙢𝙗𝙚𝙧 𝙤𝙛 𝙚𝙡𝙚𝙢𝙚𝙣𝙩𝙨, 𝙏𝙤𝙩𝙖𝙡 𝙨𝙪𝙢, 𝘼𝙫𝙚𝙧𝙖𝙜𝙚, 𝙃𝙞𝙜𝙝𝙚𝙨𝙩 𝙣𝙪𝙢𝙗𝙚𝙧, 𝙇𝙤𝙬𝙚𝙨𝙩 𝙣𝙪𝙢𝙗𝙚𝙧.


```python
numbers = [10,34,56,79,38,28,35,72,80]
def outer(numbers):
    def inner():
        high = numbers[0]
        low = numbers[0]
        count = 0
        total = 0
        
        for num in numbers:
            total += num
            count += 1
            average = total /count
            if num > high:
                high = num
            elif num < low :
                low = num
        return count,total,average,high ,low
                
    return inner()
outer(numbers)
```
