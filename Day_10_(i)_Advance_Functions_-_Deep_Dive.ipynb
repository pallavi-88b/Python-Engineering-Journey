## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟏 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙋𝙮𝙩𝙝𝙤𝙣 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙥𝙧𝙞𝙣𝙩 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙛𝙧𝙤𝙢 𝟏 𝙩𝙤 𝟓.

```python
def rec(num):
    if num > 5: # Base Case
        return  # Stop Condition
    print(num) # Function That Work
    rec(num+1) # Recursive Case
rec(1)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟐 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙥𝙧𝙞𝙣𝙩 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙛𝙧𝙤𝙢 𝟓 𝙩𝙤 𝟏.

```python
def rec(num):
    if num == 0:
        return
    print(num)
    rec(num-1)
rec(5)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟑 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚 𝙩𝙝𝙚 𝙨𝙪𝙢 𝙤𝙛 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙛𝙧𝙤𝙢 𝟏 𝙩𝙤 𝟓.

```python
def add(num):
    if num > 5:
        return 0
    return num + add(num+1)
add(1)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟒 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚 𝙩𝙝𝙚 𝙛𝙖𝙘𝙩𝙤𝙧𝙞𝙖𝙡 𝙤𝙛 𝙖 𝙣𝙪𝙢𝙗𝙚𝙧.

```python
def fact(num):
    if num == 0:
        return 1
    return num * fact(num-1)
fact(6)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟓 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚 𝙩𝙝𝙚 𝙥𝙧𝙤𝙙𝙪𝙘𝙩 𝙤𝙛 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙛𝙧𝙤𝙢 𝟏 𝙩𝙤 𝙉.

```python
def multiply(num):
    if num == 1:
        return 1
    return num * multiply(num - 1)
multiply(5)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟔 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙘𝙤𝙪𝙣𝙩 𝙛𝙧𝙤𝙢 𝟏 𝙩𝙤 𝙉 𝙖𝙣𝙙 𝙧𝙚𝙩𝙪𝙧𝙣 𝙩𝙝𝙚 𝙩𝙤𝙩𝙖𝙡 𝙣𝙪𝙢𝙗𝙚𝙧 𝙤𝙛 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙘𝙤𝙪𝙣𝙩𝙚𝙙.

```python
def rec(num):
    if num == 0:
        return 0
    return 1 + rec(num-1)
rec(5)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟕 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚 𝙩𝙝𝙚 𝙨𝙪𝙢 𝙤𝙛 𝙚𝙫𝙚𝙣 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙛𝙧𝙤𝙢 𝟏 𝙩𝙤 𝙉.

```python
def even(num):
    if num == 0:
        return 0
    if num % 2 == 0:
        return num + even(num - 1)
    else:
        return even(num-1)
even(10)
```
