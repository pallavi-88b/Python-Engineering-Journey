## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟗 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙋𝙮𝙩𝙝𝙤𝙣 𝙥𝙧𝙤𝙜𝙧𝙖𝙢 𝙪𝙨𝙞𝙣𝙜 𝙢𝙖𝙥() 𝙖𝙣𝙙 𝙖 𝙡𝙖𝙢𝙗𝙙𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚 𝙩𝙝𝙚 𝙨𝙦𝙪𝙖𝙧𝙚 𝙤𝙛 𝙤𝙣𝙡𝙮 𝙩𝙝𝙚 𝙥𝙤𝙨𝙞𝙩𝙞𝙫𝙚 𝙣𝙪𝙢𝙗𝙚𝙧𝙨.


```python
num = [-3,2,-5,4,0,6]
check = map(lambda a: a*a if a>0 else a,num)
print(list(check))
```
