## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟐𝟎 : 𝙏𝙖𝙠𝙚 𝙖 𝙡𝙞𝙨𝙩 𝙤𝙛 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙖𝙣𝙙 𝙛𝙞𝙣𝙙: 𝙏𝙤𝙩𝙖𝙡 𝙚𝙡𝙚𝙢𝙚𝙣𝙩𝙨, 𝙎𝙪𝙢, 𝘼𝙫𝙚𝙧𝙖𝙜𝙚, 𝙇𝙖𝙧𝙜𝙚𝙨𝙩, 𝙎𝙢𝙖𝙡𝙡𝙚𝙨𝙩, 𝙀𝙫𝙚𝙣 𝙘𝙤𝙪𝙣𝙩, 𝙊𝙙𝙙 𝙘𝙤𝙪𝙣𝙩, 𝙋𝙤𝙨𝙞𝙩𝙞𝙫𝙚 𝙘𝙤𝙪𝙣𝙩, 𝙉𝙚𝙜𝙖𝙩𝙞𝙫𝙚 𝙘𝙤𝙪𝙣𝙩, 𝙒𝙞𝙩𝙝𝙤𝙪𝙩 𝙪𝙨𝙞𝙣𝙜 𝙢𝙖𝙭(), 𝙢𝙞𝙣() 𝙤𝙧 𝙖𝙣𝙮 𝙨𝙝𝙤𝙧𝙩𝙘𝙪𝙩 𝙩𝙝𝙖𝙩 𝙙𝙞𝙧𝙚𝙘𝙩𝙡𝙮 𝙨𝙤𝙡𝙫𝙚𝙨 𝙩𝙝𝙚𝙨𝙚.

```python
numbers = [80,53,82,73,-73,98,-54,98,72,-65,100,-62,72]
count = 0
Sum = 0
Max = numbers[0]
Min = numbers[0]
Even_count = 0
Odd_count = 0
Positive_count = 0
Negative_count = 0
for num in numbers:
    count = count + 1
    Sum = Sum + num
    if num > 0:
        Positive_count = Positive_count + 1
    elif num < 0:
        Negative_count = Negative_count + 1
    if num % 2 == 0:
        Even_count = Even_count + 1
    elif num % 2 != 0:
        Odd_count = Odd_count + 1
    if num > Max:
        Max = num
    elif num < Min:
        Min = num
Average = Sum / count
print("Count: ",count)
print("Sum: ",Sum)
print("Positive Count: ",Positive_count)
print("Negative Count: ",Negative_count)
print("Even Count: ",Even_count)
print("Odd Count: ",Odd_count)
print("Average: ",Average)
print("Maximum Number: ",Max)
print("Minimum Number: ",Min)
```

```python
marks = [78, 45, 92, 33, 67, 89, 45, 56, 92, 71, 38, 80]

Total_Students = 0
Total_Marks = 0
Highest_Marks = marks[0]
Second_Highest_Marks = marks[0]
Lowest_Marks = marks[0]
Even_Marks_Count = 0
Odd_Marks_Count = 0
Pass_Students = []
Fail_Students = []
seen = []
duplicates = []
unique_marks = []
frequency_data = []

for mark in marks:

    Total_Students = Total_Students + 1

    Total_Marks = Total_Marks + mark

    if mark > Highest_Marks:
        Second_Highest_Marks = Highest_Marks
        Highest_Marks = mark

    elif mark > Second_Highest_Marks and mark != Highest_Marks:
        Second_Highest_Marks = mark

    if mark < Lowest_Marks:
        Lowest_Marks = mark

    if mark % 2 == 0:
        Even_Marks_Count = Even_Marks_Count + 1
    else:
        Odd_Marks_Count = Odd_Marks_Count + 1

    if mark >= 40:
        Pass_Students.append(mark)
    else:
        Fail_Students.append(mark)

    if mark in seen:
        if mark not in duplicates:
            duplicates.append(mark)
    else:
        seen.append(mark)
        unique_marks.append(mark)
Average = Total_Marks / Total_Students
for unique_mark in unique_marks:

    frequency = 0

    for mark in marks:
        if mark == unique_mark:
            frequency = frequency + 1

    frequency_data.append([unique_mark, frequency])

print("========== STUDENT MARKS ANALYZER ==========")
print("Total Students:", Total_Students)
print("Total Marks:", Total_Marks)
print("Average Marks:", Average)
print("Highest Marks:", Highest_Marks)
print("Second Highest Marks:", Second_Highest_Marks)
print("Lowest Marks:", Lowest_Marks)
print("Even Marks Count:", Even_Marks_Count)
print("Odd Marks Count:", Odd_Marks_Count)
print("Pass Students:", Pass_Students)
print("Fail Students:", Fail_Students)
print("Duplicate Marks:", duplicates)
print("Unique Marks:", unique_marks)
print("Frequency of Marks:")

for data in frequency_data:
    print(data[0], "->", data[1])
```
