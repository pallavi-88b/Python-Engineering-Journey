## Question 03 : Take radius as input from user and print Area & Circumference of circle 

```python
radius = float(input("enter the radius of the circle :"))
pi = 3.14
area = pi * radius * radius
circumference = 2 * pi * radius 
print(f'The circumference of the circle : {circumference}')
print(f'The area of the circle : {area}')
```

## Question 09 : Take basic salary as input and calculate HRA (20%) , DA (10%) , Bonus (5%) and Gross salary

## Question 17 : Take cost price and selling price as an input and calculate the profit/loss and it's percentage

# Mini project 01 : Personal expense calculator

```python
monthly_salary = float(input("Enter monthly salary ;"))
Rent= float(input("Enter rent :"))
Food= float(input("Enter food expense :"))
Travel= float(input("Enter travel expense :"))
Shopping = float(input("Enter shoopping expense :"))
other_expense = float(input("Enter other expense :"))

total = Rent + Food + Travel + Shopping + other_expense 
remaining_amount = monthly_salary - total
saving_percentage = (remaining_amount / total ) * 100

print(f'The total monthly expense : {total}')
print(f'The remaining amount : {remaining_amount}')
print(f'Saving percentage : {saving_percentage}')


```
