```python
n = int(input("Enter Number: "))

def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
        
print(factorial(n))
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟓 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙛𝙞𝙣𝙙 𝙩𝙝𝙚 𝙣𝙩𝙝 𝙁𝙞𝙗𝙤𝙣𝙖𝙘𝙘𝙞 𝙣𝙪𝙢𝙗𝙚𝙧.

```python
def fibonacci(number):
    a = 0
    b = 1
    if number == 1:
        return a
    elif number == 2:
        return a, b
    for num in range(number):
        print(a, end=" ")
        next_term = a + b
        a = b
        b = next_term
    
fibonacci(10)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟎𝟖 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙙𝙚𝙘𝙤𝙧𝙖𝙩𝙤𝙧 𝙩𝙝𝙖𝙩 𝙡𝙤𝙜𝙨 𝙩𝙝𝙚 𝙖𝙧𝙜𝙪𝙢𝙚𝙣𝙩𝙨 𝙖𝙣𝙙 𝙧𝙚𝙩𝙪𝙧𝙣 𝙫𝙖𝙡𝙪𝙚 𝙤𝙛 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣.

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print(f'Arguments: {args}')
        result = func(*args, **kwargs)
        print(f'Return value: {result}')
        return result
    return wrapper

@decorator
def add(a, b):
    return a + b

result = add(10, 20)
print("Final Result:", result)
```

```python
def apply_function(func, numbers):
    return list(map(func, numbers))

print(apply_function(lambda x: x * 2, [1, 2, 3, 4]))
```
