```python
students = {
    "Gourav": 85,
    "Rahul": 38,
    "Aman": 92,
    "Karan": 45,
    "Rohit": 33,
    "Anuj":62,
    "Harsh":41,
    "Raj":71,
    "Raman":34,
    "Kamal":26,
    "Sapna":57
}

total_students = list(students.keys())
total_marks = list(students.values())
total = 0
count = 0
highest = total_marks[0]
lowest = total_marks[0]
highest_scorer = list(students.keys())[0]
lowest_scorer = list(students.keys())[0]
pass_list = []
fail_list = []
pcount = 0
fcount = 0

for name,marks in students.items():
    count = count + 1
    total = total + marks
    if marks > highest:
        highest = marks
        highest_scorer = name
    elif marks < lowest:
        lowest = marks
        lowest_scorer = name
    if marks >= 45:
        pass_list.append(name)
        pcount = pcount + 1
    elif marks < 45:
        fail_list.append(name)
        fcount = fcount + 1
Average = total / count
print(f'Total Students: {count}')
print(f'Total Marks: {total}')
print(f'Average Marks: {Average:.2f}')
print(f'Highest Marks: {highest}')
print(f'Lowest Marks: {lowest}')
print(f'Highest Scorer Student: {highest_scorer}')
print(f'Lowest Scorer Student: {lowest_scorer}')
print(f'Pass Students List: {pass_list}')
print(f'Pass Students Count: {pcount}')
print(f'Fail Students List: {fail_list}')
print(f'Fail Students Count: {fcount}')
```
