## Question: 08 Given a tuple of numbers, count the number of positive, negative, and zero elements.



```python
number = (89,0,4 ,78 ,-39 ,- 67 ,97 ,- 23 ,0 ,2 ,0 ,5 )
positive = 0
negative = 0
zero = 0
for i in number:
    if i == 0:
        zero += 1
    elif i > 0:
        positive += 1
    else:
        negative += 1
print(f'Total positive element:{positive}')
print(f'Total zero element:{zero}')
print(f'Total negative element:{negative}')
```

## Question: 20 Given two lists:

list1 = [10, 20, 30, 40, 50, 20, 30]
list2 = [30, 40, 50, 60, 70, 30, 40]
Write a program to:

Convert both lists into unique sets.
Find the common unique elements.
Find the elements present only in list1.
Find the elements present only in list2.
Find all unique elements present in both lists.
Count the number of common elements.




```python
list1 = [10, 20, 30, 40, 50, 20, 30]
list2 = [30, 40, 50, 60, 70, 30, 40]
list_1 = set(list1)
list_2 = set(list2)
only_list1 = list_1 - list_2
only_list2 = list_2 - list_1
all_unique = list_1 | list_2

print(f'Unique list 1:{set(list_1)}')
print(f'Unique list 2:{set(list_2)}')
print(f'Common elements:{list_1 & list_2}')
print(f'Only list 1:{only_list1}')
print(f'Only list 2:{only_list2}')
print(f'All unique element:{all_unique}')




```

```python
marks = (78, 45, 92, 33, 67, 89, 45, 56, 92, 71, 38, 80)

count = 0
total = 0
Max = marks[0]
Min = marks[0]
even = 0
odd = 0
unique = set(marks)
seen = []
duplicates = []
pass_mark = []
fail_mark = []

for mark in marks:
    count = count + 1
    total = total + mark
    if mark > Max:
        Max = mark
    elif mark < Min:
        Min = mark
    if mark % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1
    if mark >= 40:
        pass_mark.append(mark)
    else:
        fail_mark.append(mark)
        
    if mark in seen:
        if mark not in duplicates:
            duplicates.append(mark)
    else:
        seen.append(mark)
        
Average = total / count
print("\nFrequency of Marks:")

for um in unique:
    frequency = 0
    for mark in marks:
        if mark == um:
            frequency = frequency + 1
    print(um, "→", frequency)

group1 = {78, 45, 92, 67, 89}
group2 = {45, 56, 92, 71, 89}

common_marks = group1 & group2

print("\n===== STUDENT PERFORMANCE REPORT =====")

print("Total Students:", count)
print("Total Marks:", total)
print("Average Marks:", Average)
print("Highest Marks:", Max)
print("Lowest Marks:", Min)

print("Even Marks:", even)
print("Odd Marks:", odd)

print("Pass Marks:", pass_mark)
print("Fail Marks:", fail_mark)

print("Unique Marks:", unique)
print("Duplicate Marks:", duplicates)

print("Group 1:", group1)
print("Group 2:", group2)
print("Common Marks:", common_marks)
print("Common Marks Count:", len(common_marks))
```
