## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟗 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙛𝙞𝙣𝙙_𝙝𝙞𝙜𝙝𝙚𝙨𝙩_𝙨𝙩𝙪𝙙𝙚𝙣𝙩(𝙨𝙩𝙪𝙙𝙚𝙣𝙩𝙨) 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙩𝙝𝙚 𝙣𝙖𝙢𝙚 𝙖𝙣𝙙 𝙢𝙖𝙧𝙠𝙨 𝙤𝙛 𝙩𝙝𝙚 𝙨𝙩𝙪𝙙𝙚𝙣𝙩 𝙬𝙞𝙩𝙝 𝙩𝙝𝙚 𝙝𝙞𝙜𝙝𝙚𝙨𝙩 𝙢𝙖𝙧𝙠𝙨.

```python
students = {
    "Pallavi": 85,
    "Rahul": 38,
    "Aman": 92,
    "Karan": 45,
    "Rohit": 33,
    "Anuj": 62,
    "Harsh": 41,
    "Raj": 71,
    "Raman": 34,
    "Kamal": 26,
    "Sapna": 57
}
def highest_student(students):
    high = list(students.values())[0]
    Max = list(students.keys())[0]
    for key, value in students.items():
        if value > high:
            high = value
            Max = key
    return Max, high
result = highest_student(students)
print(result)
```

```python
students = {
    "Pallavi": 85,
    "Rahul": 38,
    "Aman": 92,
    "Karan": 45,
    "Rohit": 33,
    "Anuj": 62,
    "Harsh": 41,
    "Raj": 71,
    "Raman": 34,
    "Kamal": 26,
    "Sapna": 57
}
def student_analysis(students):
    total = 0
    count = 0
    Max = list(students.values())[0]
    Min = list(students.values())[0]
    highest = list(students.keys())[0]
    lowest = list(students.keys())[0]
    pass_count = 0
    fail_count = 0
   
    for name,marks in students.items():
        count = count + 1
        total = total + marks
        if marks > Max:
            Max = marks
            highest = name
        elif marks < Min:
            Min = marks
            lowest = name
        if marks >= 40:
            pass_count = pass_count + 1
        elif marks < 40:
            fail_count = fail_count + 1

    Average = total / count
    return count, total, Average, Max, Min, highest, pass_count, fail_count

student_analysis(students)
```
