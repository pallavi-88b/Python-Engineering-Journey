```python
text = "PyThon123@#"
upper = 0
lower = 0
digit = 0
special = 0
for t in text:
    if t.isupper():
        upper = upper + 1
    elif t.islower():
        lower = lower + 1
    elif t.isdigit():
        digit = digit + 1
    else:
        special = special + 1

print("Uppercase: ",upper)
print("Lowercase: ",lower)
print("Digit: ",digit)
print("Special: ",special)
```

```python
s = "ADOBECODEBANC"
t = "ABC"
result = ""

for i in range(len(s)):
    for j in range(i + 1, len(s) + 1):
        sub = s[i:j]
        if all(sub.count(c) >= t.count(c) for c in t):
            if result == "" or len(sub) < len(result):
                result = sub

print(result)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟗𝟗 : 𝙃𝙤𝙬 𝙙𝙤 𝙮𝙤𝙪 𝙛𝙞𝙣𝙙 𝙩𝙝𝙚 𝙣𝙪𝙢𝙗𝙚𝙧 𝙤𝙛 𝙨𝙪𝙗𝙨𝙩𝙧𝙞𝙣𝙜𝙨 𝙩𝙝𝙖𝙩 𝙖𝙧𝙚 𝙥𝙖𝙡𝙞𝙣𝙙𝙧𝙤𝙢𝙚𝙨 𝙞𝙣 𝙖 𝙨𝙩𝙧𝙞𝙣𝙜?

```python
text = "aaa"
count = 0

for i in range(len(text)):
    for j in range(i + 1, len(text) + 1):
        sub = text[i:j]
        if sub == sub[::-1]:
            count += 1

print(count)
```
