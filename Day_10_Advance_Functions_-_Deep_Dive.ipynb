## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟕 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙛𝙞𝙣𝙙 𝙩𝙝𝙚 𝙖𝙫𝙚𝙧𝙖𝙜𝙚 𝙤𝙛 𝙖𝙡𝙡 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙞𝙣 𝙖 𝙡𝙞𝙨𝙩.

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟐𝟎 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙧𝙚𝙘𝙪𝙧𝙨𝙞𝙫𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙘𝙤𝙪𝙣𝙩 𝙝𝙤𝙬 𝙢𝙖𝙣𝙮 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙖𝙧𝙚 𝙜𝙧𝙚𝙖𝙩𝙚𝙧 𝙩𝙝𝙖𝙣 𝙩𝙝𝙚 𝙖𝙫𝙚𝙧𝙖𝙜𝙚 𝙤𝙛 𝙩𝙝𝙚 𝙡𝙞𝙨𝙩.


```python
def avg(num,idx):
    if idx == len(num):
        return 0,0
    count,total = avg(num,idx+1)
    count = count + 1
    total = total + num[idx]
    return count,total

def high(num,idx,average):
    if idx == len(num):
        return 0
    if num[idx] > average:
        return 1 + high(num, idx+1,average)
    else:
        return high(num,idx+1,average)

numbers = [10,20,30,40,50]
count,total = avg(numbers,0)
average = total/count
result = high(numbers,0,average)
print("Total:",total)
print("Count:",count)
print("Average:",average)
print("Result:",result)



```
