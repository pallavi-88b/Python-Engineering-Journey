## Question: 13 Create a function marks_analysis(marks) that returns total marks and average marks.



```python
marks = [2,4,5,6,8,8]
def marks_analysis(marks):
    total = 0
    avg = 0
    count = 0
    for mark in marks:
        total = total + mark
        count += 1
        avg = total / count
    return total,avg
marks_analysis(marks)
```

## Question: 14 Create a function list_analysis(numbers) that returns the highest and lowest number without using max() or min().



```python
numbers = [10,20,50,60,49]
def list_analysis(numbers):
    high = numbers[0]
    low = numbers[0]
    for num in numbers:
        if num > high:
            high = num
        elif num < low:
            low = num
    return high , low
list_analysis(numbers)
```

## Question: 15 Create a function count_numbers(numbers) that returns the count of positive numbers and negative numbers.

## Question: 19 Create a function student_analysis(students) that returns total students, total marks, average marks, and highest marks.



```python
students = [65,25,14,89,36,74,85,36]
def student_analysis(students):
    total = 0
    tmarks = 0
    high = students[0]
    for student in students:
        total += 1
        tmarks += student
        avg = tmarks/total
        if student > high:
            high = student
    return total,tmarks,avg,high
student_analysis(students)
```

## Question: 20 Create a function sales_analysis(sales) that returns total sales, average sales, highest sale, lowest sale, and number of sales above 1000.



```python
sales = [99,87,56,37,98,45,68,1500,1000,1800,1756,1688]
def sales_analysis(sales):
    total = 0
    count = 0
    high = sales[0]
    low = sales[0]
    num = 0
    for sale in sales :
        total += sale
        count += 1
        avg = total / count
        if sale > high:
            high = sale
        elif sale < low :
            low = sale
        elif sale > 1000:
            num += 1
    return total , avg , high , low , num
sales_analysis(sales)    
```
