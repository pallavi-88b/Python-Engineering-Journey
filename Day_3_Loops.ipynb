## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟒 : 𝙏𝙖𝙠𝙚 𝙖𝙣 𝙞𝙣𝙩𝙚𝙜𝙚𝙧 𝙖𝙨 𝙞𝙣𝙥𝙪𝙩 𝙖𝙣𝙙 𝙘𝙝𝙚𝙘𝙠 𝙬𝙝𝙚𝙩𝙝𝙚𝙧 𝙞𝙩 𝙞𝙨 𝙖𝙣 𝘼𝙧𝙢𝙨𝙩𝙧𝙤𝙣𝙜 𝙣𝙪𝙢𝙗𝙚𝙧.

```python
num = int(input("Enter Number: "))
original = num
str_num = str(num)
power = len(str_num)
armstrong = 0
while num >0:
    digit = num % 10
    armstrong = armstrong + (digit ** power)
    num = num // 10
if armstrong == original:
    print(f'{original} is Armstrong Number')
else:
    print(f'{original} is not an Armstrong Number')
```

# ᴍɪɴɪ ᴘʀᴏᴊᴇᴄᴛ : ɴᴜᴍʙᴇʀ ᴀɴᴀʟʏꜱɪꜱ ᴛᴏᴏʟ

```python
num = int(input("Enter Number: "))
str_num = str(num)
power = len(str_num)
original = num
count = 0
Sum = 0
product = 1
reverse = 0
armstrong = 0

while num > 0:
    digit = num % 10
    count = count + 1
    num = num // 10
    Sum = Sum + digit
    product = product * digit
    reverse = (reverse * 10) + digit
    armstrong = armstrong + (digit ** power)
    
print(f'Digits: {count}')
print(f'Sum: {Sum}')
print(f'Digit Product: {product}')
print(f'Digit Reverse: {reverse}')

if reverse == original:
    print("Palindrome : Yes")
else:
    print("Palindrome : No")
if armstrong == original:
    print("Armstrong : Yes")
else:
    print("Armstrong : No")
```
