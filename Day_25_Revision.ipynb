## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟏 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙩𝙝𝙚 𝙩𝙤𝙩𝙖𝙡, 𝙖𝙫𝙚𝙧𝙖𝙜𝙚, 𝙝𝙞𝙜𝙝𝙚𝙨𝙩, 𝙖𝙣𝙙 𝙡𝙤𝙬𝙚𝙨𝙩 𝙫𝙖𝙡𝙪𝙚𝙨 𝙤𝙛 𝙖 𝙡𝙞𝙨𝙩 𝙬𝙞𝙩𝙝𝙤𝙪𝙩 𝙪𝙨𝙞𝙣𝙜 𝙨𝙪𝙢(), 𝙢𝙖𝙭(), 𝙤𝙧 𝙢𝙞𝙣().

```python
numbers = [10, 25, 8, 40, 17]
def calculate(numbers):
    count = 0
    total = 0
    high = numbers[0]
    low = numbers[0]
    for num in numbers:
        count += 1
        total += num
        if num > high:
            high = num
        elif num < low:
            low = num
    average = total/count
    return total, average, high, low
calculate(numbers)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟐 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙖𝙣𝙖𝙡𝙮𝙯𝙚(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙩𝙝𝙚 𝙘𝙤𝙪𝙣𝙩 𝙤𝙛 𝙥𝙤𝙨𝙞𝙩𝙞𝙫𝙚 𝙣𝙪𝙢𝙗𝙚𝙧𝙨, 𝙣𝙚𝙜𝙖𝙩𝙞𝙫𝙚 𝙣𝙪𝙢𝙗𝙚𝙧𝙨, 𝙖𝙣𝙙 𝙯𝙚𝙧𝙤𝙚𝙨 𝙞𝙣 𝙖 𝙡𝙞𝙨𝙩.

```python
def analyze(numbers):
    positive = 0
    negative = 0
    zero = 0
    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        elif num == 0:
            zero += 1
    return positive, negative, zero
analyze([10, -5, 0, 8, -2, 0, 15, -7])
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟑 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙛𝙞𝙣𝙙_𝙝𝙞𝙜𝙝𝙚𝙨𝙩(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙩𝙝𝙚 𝙝𝙞𝙜𝙝𝙚𝙨𝙩 𝙣𝙪𝙢𝙗𝙚𝙧 𝙛𝙧𝙤𝙢 𝙖 𝙡𝙞𝙨𝙩 𝙬𝙞𝙩𝙝𝙤𝙪𝙩 𝙪𝙨𝙞𝙣𝙜 𝙢𝙖𝙭().

```python
def find_highest(numbers):
    high = numbers[0]
    for num in numbers:
        if num > high:
            high = num
    return high
find_highest([12, 45, 7, 89, 34, 56])
```

```python
def filter_even(numbers):
    even = []
    for num in numbers:
        if num % 2 == 0:
            even.append(num)
    return even
filter_even([12, 7, 4, 9, 18, 25, 30])
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟐𝟎 : 𝙒𝙧𝙞𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙖𝙣𝙖𝙡𝙮𝙯𝙚𝙨 𝙖 𝙡𝙞𝙨𝙩 𝙖𝙣𝙙 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙞𝙩𝙨 𝙘𝙤𝙪𝙣𝙩, 𝙨𝙪𝙢, 𝙖𝙫𝙚𝙧𝙖𝙜𝙚, 𝙝𝙞𝙜𝙝𝙚𝙨𝙩, 𝙡𝙤𝙬𝙚𝙨𝙩, 𝙚𝙫𝙚𝙣 𝙘𝙤𝙪𝙣𝙩, 𝙥𝙤𝙨𝙞𝙩𝙞𝙫𝙚 𝙘𝙤𝙪𝙣𝙩, 𝙣𝙚𝙜𝙖𝙩𝙞𝙫𝙚 𝙘𝙤𝙪𝙣𝙩, 𝙯𝙚𝙧𝙤 𝙘𝙤𝙪𝙣𝙩, 𝙖𝙣𝙙 𝙪𝙣𝙞𝙦𝙪𝙚 𝙫𝙖𝙡𝙪𝙚𝙨.

```python
def analyze(numbers):
    count = 0
    add = 0
    high = numbers[0]
    low = numbers[0]
    even = 0
    positive = 0
    negative = 0
    zero = 0
    unique = []
    for num in numbers:
        count += 1
        add += num
        if num not in unique:
            unique.append(num)
        if num > high:
            high = num
        elif num < low:
            low = num
        if num % 2 == 0:
            even += 1
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        elif num == 0:
            zero += 1
    return count, add, high, low, even, positive, negative, zero, unique
analyze([10, -5, 20, 10, 0, -8, 25, 20, 7])
```
