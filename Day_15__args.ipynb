```python
def number(*args):
    total = 0
    add =0 
    high = args[0]
    low = args[0]
    for num in args:
        total += 1
        add += num
        if num > high:
            high = num
        elif num< low:
            low = num
    return total,add,high,low

number(2,8,34,25,16,19)
```
